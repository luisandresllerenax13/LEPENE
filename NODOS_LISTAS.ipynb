```python
class MyList:
    def __init__(self, initial_items=None):
        self._data = []
        if initial_items is not None:
            for item in initial_items:
                self._data.append(item)

    def size(self):
        """Returns the number of elements in the list."""
        return len(self._data)

    def isempty(self):
        """Returns True if the list is empty, False otherwise."""
        return len(self._data) == 0

    def setsize(self, new_size, fill_value=None):
        """Resizes the list to new_size. If new_size is larger, fills new slots with fill_value. If smaller, truncates the list."""
        current_size = len(self._data)
        if new_size > current_size:
            self._data.extend([fill_value] * (new_size - current_size))
        elif new_size < current_size:
            self._data = self._data[:new_size]

    def first(self):
        """Returns the first element of the list. Raises IndexError if the list is empty."""
        if self.isempty():
            raise IndexError("List is empty")
        return self._data[0]

    def last(self):
        """Returns the last element of the list. Raises IndexError if the list is empty."""
        if self.isempty():
            raise IndexError("List is empty")
        return self._data[-1]

    def __str__(self):
        return str(self._data)

    def __repr__(self):
        return repr(self._data)

# Example Usage:
my_list = MyList([10, 20, 30, 40, 50])
print(f"Initial list: {my_list}")
print(f"Size: {my_list.size()}")
print(f"Is empty: {my_list.isempty()}")
print(f"First element: {my_list.first()}")
print(f"Last element: {my_list.last()}")

my_list.setsize(3) # Truncate
print(f"List after setsize(3): {my_list}")
print(f"Size: {my_list.size()}")

my_list.setsize(7, fill_value=None) # Extend with None
print(f"List after setsize(7, None): {my_list}")
print(f"Size: {my_list.size()}")

my_list.setsize(0) # Empty the list
print(f"List after setsize(0): {my_list}")
print(f"Is empty: {my_list.isempty()}")

try:
    my_list.first()
except IndexError as e:
    print(f"Error accessing first element on empty list: {e}")
```

```python
import random

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def append(self, data):
        """Adds a new node with data to the end of the list."""
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node
        self.size += 1

    def remove_first(self):
        """Removes and returns the first element. Returns None if empty."""
        if not self.head:
            return None

        removed_data = self.head.data
        self.head = self.head.next
        self.size -= 1

        if self.head is None:
            self.tail = None

        return removed_data

    def remove_last(self):
        """Removes and returns the last element. Returns None if empty."""
        if not self.head:
            return None

        removed_data = self.tail.data

        if self.head == self.tail:
            self.head = None
            self.tail = None
        else:
            current = self.head
            while current.next != self.tail:
                current = current.next
            current.next = None
            self.tail = current

        self.size -= 1
        return removed_data

    def __str__(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        return " -> ".join(elements) if elements else "Empty List"

# Create a list and populate it with random data
my_linked_list = LinkedList()
print("Populating linked list with 5 random integers:")
for _ in range(5):
    rand_val = random.randint(10, 99)
    my_linked_list.append(rand_val)
    print(f"Appended: {rand_val}")

print(f"\nCurrent List: {my_linked_list}")
print(f"Head: {my_linked_list.head.data if my_linked_list.head else None}")
print(f"Tail: {my_linked_list.tail.data if my_linked_list.tail else None}")

# Remove first
removed_val = my_linked_list.remove_first()
print(f"\nRemoved first: {removed_val}")
print(f"List after remove_first(): {my_linked_list}")
print(f"New Head: {my_linked_list.head.data if my_linked_list.head else None}")
print(f"New Tail: {my_linked_list.tail.data if my_linked_list.tail else None}")

# Remove last
removed_val = my_linked_list.remove_last()
print(f"\nRemoved last: {removed_val}")
print(f"List after remove_last(): {my_linked_list}")
print(f"New Head: {my_linked_list.head.data if my_linked_list.head else None}")
print(f"New Tail: {my_linked_list.tail.data if my_linked_list.tail else None}")
```

```python
class Arraylista:
  def __init__(self):
    self.__data = MyList()

  def push(self, element):

    current_size = self.__data.size()
    self.__data.setsize(current_size + 1)
    self.__data._data[current_size] = element

  def pop(self):
    if self.is_empty():
      raise IndexError("Pop from empty stack")
    last_element = self.__data.last()
    self.__data.setsize(self.__data.size() - 1)
    return last_element

  def size(self):
    return self.__data.size()

  def is_empty(self):
    return self.__data.isempty()

  def __str__(self):
    return str(self.__data)
```

```python
stack = Arraylista()
print(f"Initial stack is empty? {stack.is_empty()}")

print("\n--- Pushing elements: 10, 20, 30 ---")
stack.push(10)
stack.push(20)
stack.push(30)
print(f"Stack contents: {stack}")
print(f"Size after pushes: {stack.size()}")
print(f"Is empty? {stack.is_empty()}")

print("\n--- Popping elements ---")
print(f"Popped: {stack.pop()}")
print(f"Stack after 1 pop: {stack}")
print(f"Popped: {stack.pop()}")
print(f"Popped: {stack.pop()}")

print(f"\nStack is empty now? {stack.is_empty()}")
print(f"Final size: {stack.size()}")
```
